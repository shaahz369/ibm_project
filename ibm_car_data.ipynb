```python
# Scatter plot with annotation
plt.figure(figsize=(8,6))
sns.scatterplot(x=df['mpg'], y=df['cyl'])
plt.title('Scatter Plot between mpg and cyl')
plt.xlabel('mpg')
plt.ylabel('cyl')

# Example of adding a description for a specific point
plt.annotate('This point represents the highest value', 
             xy=(df['mpg'][100], df['cyl'][100]), 
             xytext=(df['mpg'][100] + 1, df['cyl'][100] + 1),
             arrowprops=dict(arrowstyle="->"))

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv('your_file.csv')

# Scatter plot with annotation using only matplotlib
plt.figure(figsize=(8, 6))

# Create scatter plot between 'mpg' and 'cyl'
plt.scatter(df['mpg'], df['cyl'])

# Add title and labels
plt.title('Scatter Plot between MPG and Cylinders')
plt.xlabel('MPG (Miles per Gallon)')
plt.ylabel('Cylinders')

# Example of adding a description for a specific point
plt.annotate('This point represents the highest value', 
             xy=(df['mpg'][100], df['cyl'][100]), 
             xytext=(df['mpg'][100] + 1, df['cyl'][100] + 1),
             arrowprops=dict(arrowstyle="->"))

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv('your_file.csv')

# Subplots with multiple graphs using matplotlib
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Histogram for 'mpg' (replace 'column_name' with your actual column)
axes[0, 0].hist(df['mpg'], bins=20, edgecolor='black')
axes[0, 0].set_title('Distribution of MPG')
axes[0, 0].set_xlabel('MPG (Miles per Gallon)')
axes[0, 0].set_ylabel('Frequency')

# Boxplot for 'mpg'
axes[0, 1].boxplot(df['mpg'])
axes[0, 1].set_title('Boxplot of MPG')
axes[0, 1].set_xlabel('MPG (Miles per Gallon)')

# Bar plot for cylinder count ('cyl')
category_counts = df['cyl'].value_counts()
axes[1, 0].bar(category_counts.index, category_counts.values)
axes[1, 0].set_title('Count of Cylinders')
axes[1, 0].set_xlabel('Number of Cylinders')
axes[1, 0].set_ylabel('Count')
axes[1, 0].tick_params(axis='x', rotation=45)

# Scatter plot between 'mpg' and 'cyl'
axes[1, 1].scatter(df['mpg'], df['cyl'])
axes[1, 1].set_title('Scatter Plot between MPG and Cylinders')
axes[1, 1].set_xlabel('MPG (Miles per Gallon)')
axes[1, 1].set_ylabel('Cylinders')

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```
